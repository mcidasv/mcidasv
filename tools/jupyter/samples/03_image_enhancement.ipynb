## Enhancement + annotation + a colorbar on the captured image

```python
%%mcv --capture none
import os
imageDir = expandpath('~')

ir_data = loadADDEImage(server='adde.ucar.edu', dataset='RTGOESR',
                        descriptor='M1C14', unit='TEMP', size='ALL', mag=(1, 1))

panel = buildWindow(height=600, width=800)
layer = panel[0].createLayer('Image Display', ir_data)
panel[0].setWireframe(False)

layer.setEnhancement('ABI IR Temperature', range=(200, 320))

panel[0].annotate('GOES-R Clean IR (Band 14)', line=40, element=320,
                  size=18, color='white')

colorbar = dict(
    Colorbar=dict(
        display=layer,
        width=300, height=20,
        orientation='bottom',
        tickMarks=5,
        place='LM,0,-30',
    ),
)
outFile = os.path.join(imageDir, 'G16-IR-colorbar.png')
panel[0].captureImage(outFile, width=800, height=600, formatting=[colorbar])
```
