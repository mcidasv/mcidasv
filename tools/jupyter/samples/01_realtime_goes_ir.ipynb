## Request, display, enhance, capture

```python
%%mcv --size 640x480
adde_parms = dict(
    debug=True,
    server='adde.ucar.edu',
    dataset='RTGOESR',
    size='ALL',
    mag=(1, 1),
    unit='TEMP',
)

ir_data = loadADDEImage(descriptor='M1C14', **adde_parms)

ir_label = '%s %s' % (ir_data['sensor-type'], ir_data['nominal-time'])

panel = buildWindow(height=600, width=800)
ir_layer = panel[0].createLayer('Image Display', ir_data)

panel[0].setWireframe(False)
ir_layer.setLayerLabel(label=ir_label, size=14)
ir_layer.setEnhancement('ABI IR Temperature', range=(220, 300))
```
