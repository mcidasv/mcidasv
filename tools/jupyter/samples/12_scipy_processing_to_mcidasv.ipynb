## A noisy synthetic field

```python
import numpy as np
import mcidasv_jupyter as mcv
session = mcv.get_session()
rng = np.random.default_rng(0)

lats = np.linspace(20, 52, 200)
lons = np.linspace(-125, -66, 360)
LON, LAT = np.meshgrid(lons, lats)
clean = 280 + 20*np.exp(-(((LAT-38)**2 + (LON-96+0*LON)**2)/(2*7.0**2)))
noisy = (clean + rng.normal(0, 3, clean.shape)).astype('f4')
```

## Smooth it with scipy and display

```python
from scipy.ndimage import gaussian_filter
smoothed = gaussian_filter(noisy, sigma=3).astype('f4')

session.run('''
panel = buildWindow(height=600, width=800)
layer = panel[0].createLayer('Color-Shaded Plan View', g)
panel[0].setProjection('US>CONUS')
layer.setLayerLabel(label='scipy gaussian_filter(sigma=3)')
''', arrays={'g': (smoothed, lats, lons)})
```

## Edge magnitude (Sobel) with scipy and display

```python
from scipy.ndimage import sobel
gx = sobel(smoothed, axis=1)
gy = sobel(smoothed, axis=0)
edges = np.hypot(gx, gy).astype('f4')

session.run('''
panel = buildWindow(height=600, width=800)
layer = panel[0].createLayer('Color-Shaded Plan View', g)
panel[0].setProjection('US>CONUS')
layer.setLayerLabel(label='scipy Sobel edge magnitude')
''', arrays={'g': (edges, lats, lons)})
```
