# Animating a numpy time series in McIDAS-V

```python
%load_ext mcidasv_jupyter
%mcv_connect /path/to/runMcV
%mcv_replay off
```

## Generate a propagating wave as a stack of frames

```python
import numpy as np
import mcidasv_jupyter as mcv
session = mcv.get_session()

lats = np.linspace(20, 52, 160)
lons = np.linspace(-125, -66, 300)
LON, LAT = np.meshgrid(lons, lats)
nt = 8
frames = np.stack([
    280 + 15*np.sin(np.radians(4*(LON + 8*t))) * np.cos(np.radians(3*LAT))
    for t in range(nt)
]).astype('f4')
print('cube shape (time, lat, lon):', frames.shape)
```

## Push the cube and animate it

```python
import os
OUTDIR = 'output'
os.makedirs(OUTDIR, exist_ok=True)
movie = os.path.join(OUTDIR, 'wave.gif')

session.animate_grid(frames, lats, lons, name='wave', out=movie, fps=5,
                     setup="panel[0].annotate('numpy wave', line=470, element=360, size=16, color='white')")
```
